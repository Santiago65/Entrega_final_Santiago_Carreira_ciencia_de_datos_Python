## 🔹 4- Introducción a Pandas: realizar un análisis exploratorio inicial de los DataFrames.

```python
# ============================================
#  ANÁLISIS EXPLORATORIO INICIAL
# ============================================

# Analizamos estructura y contenido de cada DataFrame

def exploracion(df, nombre):
    print(f"\n### Análisis del DataFrame: {nombre}")
    display(df.head(3))
    print("\nInformación general:")
    print(df.info())
    print("\nEstadísticas descriptivas:")
    display(df.describe(include="all"))
    print("\nCantidad de filas y columnas:", df.shape)
    print("\nTipos de datos:")
    print(df.dtypes)
    print("-" * 60)

exploracion(ventas, "Ventas")
exploracion(clientes, "Clientes")
exploracion(marketing, "Marketing")



```

## 🔹 5- Calidad de Datos: Identificar valores nulos y duplicados en los conjuntos de datos.

```python
# ============================================
#  CALIDAD DE LOS DATOS
# ============================================

# Identificar valores nulos y duplicados

def calidad(df, nombre):
    print(f"\n### Análisis de calidad: {nombre}")
    print("Valores nulos por columna:")
    display(df.isna().sum().to_frame("Nulos"))
    print("Filas duplicadas:", df.duplicated().sum())

calidad(ventas, "Ventas")
calidad(clientes, "Clientes")
calidad(marketing, "Marketing")

```


## 🔹 6- **Limpieza del dataset**



```python

# ============================================
#  LIMPIEZA DE DATOS
# ============================================

# Crear copias para limpieza
ventas_clean = ventas.copy()
clientes_clean = clientes.copy()
marketing_clean = marketing.copy()

# Eliminar duplicados
ventas_clean.drop_duplicates(inplace=True)
clientes_clean.drop_duplicates(inplace=True)
marketing_clean.drop_duplicates(inplace=True)

# Limpieza de texto en columnas tipo string
for df in [ventas_clean, clientes_clean, marketing_clean]:
    for col in df.select_dtypes(include="object").columns:
        df[col] = df[col].str.strip()

# Mostrar cambios realizados
print("Duplicados eliminados y textos normalizados.")
```

```python
# -------------------------------------------------
# 7️⃣ Guardar los DataFrames limpios como CSV
# -------------------------------------------------
#print(ventas_clean.head())
#print(clientes_clean.head())
#print(marketing_clean.head())
ventas_clean.info()
ventas_clean.to_csv("/content/drive/MyDrive/datasets/ventas_clean.csv", index=False)
clientes_clean.to_csv("/content/drive/MyDrive/datasets/clientes_clean.csv", index=False)
marketing_clean.to_csv("/content/drive/MyDrive/datasets/marketing_clean.csv", index=False)

print("✅ Archivos guardados: ventas_clean.csv, clientes_clean.csv, marketing_clean.csv")
```

##🔹 7- **Transformación de datos**

```python
# ============================================
# TRANSFORMACIÓN DE DATOS
# ============================================

# Clean and convert 'precio' to numeric
ventas_clean['precio'] = ventas_clean['precio'].astype(str).str.replace('$', '', regex=False)
ventas_clean['precio'] = pd.to_numeric(ventas_clean['precio'], errors='coerce')

# Create a column with the total of venta
ventas_clean["total_venta"] = ventas_clean["precio"] * ventas_clean["cantidad"]

# Calcular ventas totales por producto
ventas_por_producto = ventas_clean.groupby("producto")["total_venta"].sum().reset_index()

# Calcular el percentil 80
umbral_80 = ventas_por_producto["total_venta"].quantile(0.8)
print(f"Percentil 80 del total de ventas: {umbral_80:.2f}")

# Filtrar productos de alto rendimiento
productos_alto_rendimiento = ventas_por_producto.query("total_venta > @umbral_80")
print("\nProductos con alto rendimiento (superan el percentil 80):")
display(productos_alto_rendimiento)
```

##🔹 8- **Agregación**

```python
# ============================================
#  ETAPA 2.2: AGREGACIÓN DE DATOS
# ============================================

# Agrupar ventas por categoría y obtener métricas
resumen_categorias = (
    ventas_clean.groupby("categoria")
    .agg(
        total_ingresos=("total_venta", "sum"),
        promedio_precio=("precio", "mean"),
        cantidad_vendida=("cantidad", "sum")
    )
    .reset_index()
)

print("Resumen de ventas por categoría:")
display(resumen_categorias)
```

```python
# ============================================
# AGREGACIÓN DE DATOS
# Ventas por categoría de producto
# ============================================

# Convert 'precio' to numeric, handling the '$' sign and potential non-numeric values.
# This step is necessary because 'precio' was loaded as an object (string).
ventas['precio'] = ventas['precio'].astype(str).str.replace('$', '', regex=False)
ventas['precio'] = pd.to_numeric(ventas['precio'], errors='coerce') # 'coerce' will turn unparseable values into NaN

# Now, calculate 'ingreso'. Rows with NaN in 'precio' or 'cantidad' will result in NaN 'ingreso'.
ventas['ingreso'] = ventas['precio'] * ventas['cantidad']

agg_categoria = ventas.groupby("categoria").agg(
    ventas_totales = ("cantidad", "sum"),
    ingresos_totales = ("ingreso", "sum"),
    precio_promedio = ("precio", "mean")
).reset_index()

print("Resumen de ventas por categoría:")
display(agg_categoria)
```

```python
ventas_clean.dropna(inplace=True)
print("Rows with null values dropped from ventas_clean.")
print("\nInformación general después de eliminar nulos:")
display(ventas_clean.info())
print("\nValores nulos después de eliminar nulos:")
display(ventas_clean.isna().sum().to_frame("Nulos"))
```

##🔹 9- **Integración de datos**

```python
# ============================================
#  INTEGRACIÓN DE DATOS
# ============================================

# Verificar columnas comunes entre ventas y marketing
columnas_comunes = set(ventas_clean.columns).intersection(set(marketing_clean.columns))
print("Columnas comunes entre ventas y marketing:", columnas_comunes)

# Si existe 'producto', realizar integración
if "producto" in columnas_comunes:
    ventas_marketing = pd.merge(
        ventas_clean,
        marketing_clean,
        on="producto",
        how="left"
    )
    print("Integración completada. Vista previa:")
    display(ventas_marketing.head())
else:
    print("No se encontró una columna común para integrar ambos datasets.")
```

```python
# ============================================
# INTEGRACIÓN DE DATOS
# ============================================

# Unimos ventas + clientes
# La columna 'id_cliente' no existe en el DataFrame 'ventas'.
# Para poder unir estos DataFrames, 'ventas' necesitaría tener una columna que identifique al cliente.
# ventas_clientes = ventas.merge(clientes, on="id_cliente", how="left")

# Unimos ventas + marketing
ventas_marketing = ventas.merge(marketing, on="producto", how="left")

# Unimos todos (este paso dependerá de la correcta unión entre ventas y clientes)
# df_integrado = ventas_clientes.merge(marketing, on="producto", how="left")

print("Vista previa de la tabla integrada de ventas y marketing:")
display(ventas_marketing.head())
```
